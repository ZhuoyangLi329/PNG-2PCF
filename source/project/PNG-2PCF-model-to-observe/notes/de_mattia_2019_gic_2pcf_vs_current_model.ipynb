> **一句话结论：我们当前的 parameter-dependent W2 修正正确计算了窗口平均密度方差 C=σW²，但模型 ξ−C 是论文完整结果 ξ−2B(s)+C 的单项近似；Task4.3 还因为从数据重采样 random redshift 而额外包含了尚未建模的 radial integral constraint（RIC）。**

## 0. 先不要被公式吓到：问题是怎么产生的？

真实宇宙有一个真正的平均密度，但有限巡天并不知道它。我们只能用巡天内部的数据估计平均密度。因此，巡天内部看见的平均过密度会被强制设为零。

$$
\bar\delta_W \equiv \int d^3y\,W_N(\mathbf y)\delta(\mathbf y),
\qquad
\delta_{\rm obs}(\mathbf x)=\delta(\mathbf x)-\bar\delta_W,
$$

其中 $W_N$ 是归一化后的 survey selection function，满足 $\int W_N=1$。

逻辑流程可以写成：

$$
\text{真实场 }\delta
\;\longrightarrow\;
\text{用本巡天估计平均值 }\bar\delta_W
\;\longrightarrow\;
\delta_{\rm obs}=\delta-\bar\delta_W
\;\longrightarrow\;
\text{二点函数出现 4 项。}
$$

## 1. 最关键的一步：把两个括号乘开

观测二点函数来自

$$
\left\langle
[\delta_1-\bar\delta_W]
[\delta_2-\bar\delta_W]
\right\rangle.
$$

乘开以后是

$$
\underbrace{\langle\delta_1\delta_2\rangle}_{\xi(s)}
-\underbrace{\langle\delta_1\bar\delta_W\rangle}_{B_1(s)}
-\underbrace{\langle\bar\delta_W\delta_2\rangle}_{B_2(s)}
+\underbrace{\langle\bar\delta_W^2\rangle}_{C}.
$$

四项的直观含义是：

1. $\xi(s)$：真实密度与真实密度的相关；
2. $B_1(s)$：第一个点与巡天平均密度的相关；
3. $B_2(s)$：巡天平均密度与第二个点的相关；
4. $C$：巡天平均密度与自身的相关，也就是 $\sigma_W^2$。

对偶数多极矩和对称 pair average，通常有 $B_1(s)=B_2(s)\equiv B(s)$，所以完整结果是

$$
\boxed{\xi_{\rm full}(s)=\xi(s)-2B(s)+C.}
$$

## 2. 论文中的公式和上面的四项一一对应

de Mattia & Ruhlmann-Kleider (2019) 的式 (2.20) 写成

$$
\xi_\ell^{\rm cic}(s)
=
\xi_\ell^{\rm c}(s)
-IC_\ell^{\delta,{\rm ic}}(s)
-IC_\ell^{{\rm ic},\delta}(s)
+IC_\ell^{{\rm ic},{\rm ic}}(s).
$$

| 论文记号 | 直白含义 | 上一节记号 |
|---|---|---|
| $\xi^{\rm c}$ | 普通相关函数乘 survey window | $\xi$ 对应的 windowed 项 |
| $-IC^{\delta,{\rm ic}}$ | 密度与被减掉的平均模式交叉相关 | $-B_1(s)$ |
| $-IC^{{\rm ic},\delta}$ | 反方向的交叉相关 | $-B_2(s)$ |
| $+IC^{{\rm ic},{\rm ic}}$ | 被减掉的平均模式的自相关 | $+C$ |

论文这里首先计算的是尚未除以 RR 的 window-multiplied correlation。与 Landy–Szalay estimator 比较时，还要除以相同的 $RR(s)$ window。对于真实空间、各向同性 monopole 和很窄的 separation bin，普通的两点 window 大部分会被这个除法消掉；但两个 density–IC 交叉项不会因此自动消失。

## 3. 我们当前算的是什么？

我们当前的模型是

$$
\boxed{\xi_{\rm ours}(s)=\xi(s)-\sigma_W^2(\theta)=\xi(s)-C.}
$$

把它与完整结果比较：

$$
\begin{aligned}
\text{完整：}&\quad \xi-2B(s)+C,\\
\text{当前：}&\quad \xi-C.
\end{aligned}
$$

两者只有在

$$
B(s)\simeq C
$$

时才近似相等，因为此时 $-2B+C\simeq-C$。

所以，我们不是完全没有考虑交叉项的物理作用，而是用一个额外假设，把完整的 $-2B(s)+C$ 整体近似为 $-C$。问题在于：$B(s)$ 一般依赖 separation 和 survey geometry，不一定总等于常数 $C$。

### 3.1 为什么这个近似不是凭空来的？

论文的式 (2.23) 回顾了传统单项近似：

$$
P^{\rm cic}(\mathbf k)
=
P^{\rm c}(\mathbf k)
-P^{\rm c}(\mathbf 0)|\widetilde W(\mathbf k)|^2.
$$

作者明确说明：这个旧公式只使用一个有效修正项，而完整推导有三个修正项。他们也发现，在论文展示的 CMASS-like global-IC 几何中，这个单项近似与完整结果非常接近。

因此正确判断不是“我们的公式毫无依据”，而是：

> **我们的公式是有文献依据的 leading/single-term approximation，但还没有证明它在我们的 cube、positive-octant lightcone、local-PNG 和 50–350 Mpc/h 范围内足够准确。**

## 4. W2(k) 到底算对了什么？

我们的 random-pair kernel 是

$$
W_2(k)
\simeq
\frac{\sum_a RR_a\,\overline{j_0}(k;s_a)}{\sum_a RR_a},
$$

然后计算

$$
C(\theta)=\sigma_W^2(\theta)
=
\int\frac{k^2dk}{2\pi^2}P(k;\theta)W_2(k),
$$

或在 simulation mother-box 中使用相同公式的 FullDiscrete mode sum。

这一部分有三个优点：

1. $C$ 不是随手拟合的常数，而是由理论 $P(k;\theta)$ 与实际 geometry 决定；
2. 它随 $f_{\rm NL}$、$b_1$ 等参数更新；
3. Task4.2 使用解析 cubic window，Task4.3 使用 FKP-weighted random pairs。

但是，一个只依赖 $k$ 的一维 $W_2(k)$ 只能直接给出 $C$。完整的 $B(s)$ 还需要包含 pair separation $s$、内部积分距离 $\Delta$ 和 survey geometry 的更高维 window kernel；这就是当前缺失的信息。

## 8. Task4.3：还有一个更明确的 RIC 问题

Task4.3 的 random catalog 不是使用独立、预先给定的平滑 $n(z)$。它从每个 phase 的 halo redshift 中有放回重采样：

$$
\text{random radial policy}=\text{data redshift resample}.
$$

在 de Mattia & Ruhlmann-Kleider (2019) 的术语中，这就是 shuffled scheme。它不仅强制整个 survey 的平均密度为零，还会在每个径向层中压低由数据估计出来的 radial modes，因此引入 radial integral constraint：

$$
\xi_{\rm RIC}
=
\xi^{\rm c}
-IC^{\delta,{\rm rad}}
-IC^{{\rm rad},\delta}
+IC^{{\rm rad},{\rm rad}}.
$$

当前的一维 global $W_2(k)$ 没有 $\epsilon_{\rm rad}(r,x)$ 这个“是否处于同一 radial bin”的信息，所以不能表示完整 RIC。

## 10. 常见问题

### Q1：所以我们以前的结果错了吗？

不能直接说错。当前结果正确回答了“如果采用 parameter-dependent single-term GIC approximation，posterior 是什么”。问题是它还没有回答“采用论文完整三项 GIC，posterior 是否仍然相同”。

### Q2：既然 W2 来自真实 random，为什么还不完整？

因为 RR pair separation distribution 足以计算平均模式的方差 $C$，但 density–mean cross term $B(s)$ 还需要知道：当一对点被固定在 separation $s$ 时，其中一个点与整个 survey 其他位置如何相关。这是依赖 $s$ 的三点 window 信息，不包含在一条 $W_2(k)$ 曲线中。

### Q3：Landy–Szalay 不是已经除掉 survey geometry 了吗？

它用 RR 很好地消除了普通边界 pair loss，但平均密度仍然是从有限数据估计的，所以 integral constraint 不会因为除以 RR 自动消失。

### Q4：为什么论文说单项近似很好，我们还要检查？

论文证明它在特定 CMASS-like case 中数值上很好，而不是证明它对任意 geometry、任意尺度和 local PNG 都严格成立。我们的 L1000 cube、窄 radial lightcone 和低-k PNG sensitivity 都与那个示例不同。

## 14. 2026-07-12 更新：路线 A 的 radial single-term RIC 已实现

前面的第 11--12 节记录的是实现前的风险判断。本节记录现在真正采用并跑完的口径，优先级更高。

我们**没有**改测量、FKP、covariance、fit range 或 mother-box support，也**没有**声称实现论文完整三项 RIC。当前模型固定为

$$
\boxed{\xi_0^{\rm single,RIC}(s;\theta)=\xi_0^{\rm noIC}(s;\theta)-IC_0^{\rm rad,rad}(s;\theta)}.
$$

其中两个 density--RIC cross terms 仍然省略；radial normalization 已自动包含 global normalization，所以不再额外减一次 $\sigma_W^2$。$P(k)$ 侧同样只加入这个 radial response，不另加 standalone GIC：

$$
\mathbf P_{\rm obs}=\mathbf W_{\rm geom}\mathbf P_{\rm theory}-\mathbf W_{\rm RIC}\mathbf P_{\rm theory}.
$$

### 为什么最终没有继续用朴素 quadruplet MC？

最初的四点 MC 被极少数很近的 inner pair 主导，50k 到 200k 仍会出现重尾噪声。Task4.3 random 的生成式严格把 data-resampled radial redshift 与 uniform positive-octant direction 独立抽样，所以选择函数可分离：

$$W(r,\Omega)=W_{\rm rad}(r)W_{\rm octant}(\Omega).$$

因此同一个四点 radial-auto kernel 可无偏地压缩为

$$p(s,\Delta)=\sum_{a\le b}p_{ab}H_{ab}(s)H_{ab}(\Delta),$$

这里 $a,b$ 是 radial bins，$H_{ab}$ 由一次 $2^{22}$ Sobol 正八分体角向积分得到；outer marginal 再严格匹配 pycorr weighted RR。这个 factorization 去掉了重尾 MC 噪声，但没有改变 single-term RIC 的物理定义。2PCF 与 $P(k)$ 都从这一个 cache 投影。

## 15. 怎样解释最终结果？

1. radial correction 确实不是常数：在 fiducial $f_{\rm NL}=0,b_1=2.5$ 下，2PCF correction 从约 $1.20\times10^{-4}$ 随 $s$ 下降到 $2.01\times10^{-5}$。
2. 但它相对当前单份 lightcone 误差较小：逐 bin 最大约 $0.072\sigma$（2PCF）和 $0.109\sigma$（$P(k)$）。
3. 2PCF 中 no IC、旧 global single-term、新 radial single-term 分别给出 $f_{\rm NL}=3.87,7.15,10.18$ 的中位数；$P(k)$ geometry-only 与 radial single-term 分别为 $-3.86,-0.54$。所有结果仍与真值 0 相容，2PCF--$P(k)$ 也相容。
4. $\Delta\chi=1/2/4$、50k 到 200k、代表 phase、global limit、Hankel/direct、$f_{\rm NL}=0,\pm100$ 与 free-$sn_0$ gate 全部通过。
5. phase-dependent RIC correction 引入的额外 variance 远小于现有 covariance，因此本轮继续固定 covariance；这并不消除原 analytic covariance 自身已有的 diagnostic caveat。

最重要的表述边界仍然是：**这是经过收敛验证的 radial single-term approximation，不是 de Mattia & Ruhlmann-Kleider (2019) 的完整三项 RIC。**
